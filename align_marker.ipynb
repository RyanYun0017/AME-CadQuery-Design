```python
import cadquery as cq
from cadquery import Workplane, Assembly

from ocp_vscode import show, show_object, reset_show, set_port, set_defaults, get_defaults

set_port(3939)
um = 0.001
```

```python
def make_align_marker(bar_length, bar_width, thickness):
    bar_length *= um
    bar_width *= um
    thickness *= um
    wp = cq.Workplane("XY")
    # Horizontal bar
    bar1 = (wp.rect(bar_length, bar_width, centered=True).extrude(thickness))
    # Vertical bar
    bar2 = (wp.rect(bar_width, bar_length, centered=True).extrude(thickness))
    align_marker = bar1.union(bar2)
    return align_marker
```

```python
reset_show()
marker = make_align_marker(1000,100,20)
show_object(marker, name="Align Marker", options={"color": "blue"})
```

```python
# Parameters for the cross marker
l = 1.0  # length in mm
w = 0.2  # width in mm
um = 0.02  # thickness in mm (20 um)

import cadquery as cq

# Create the horizontal bar
bar1 = cq.Workplane("XY").box(l, w, um)
# Create the vertical bar and combine
bar2 = cq.Workplane("XY").box(w, l, um)
cross_marker = bar1.union(bar2)

# Show the result
show(cross_marker)
```
